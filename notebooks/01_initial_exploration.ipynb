```python
# Define the path2 to our second data file
path_2 = r"C:\Dell\Project Github\mexican-real-estate-analysis\data\raw\mexico-real-estate-2.csv"
```

```python
# Define the path3 to our third data file
path_3 = r"C:\Dell\Project Github\mexican-real-estate-analysis\data\raw\mexico-real-estate-3.csv"
```

```python
# Cleaning process using method chaining for the second dataset
df2 = (
    pd.read_csv(path_2)
    # Create "price_usd" column (19 pesos to the dollar in 2014)
    .assign(price_usd=lambda x: x['price_mxn'] / 19)
    
    # Drop "price_mxn" column
    .drop(columns=['price_mxn'])

    # Drop null values
    .dropna()
    
)
```

```python
# Print Object type, shape, and head of the cleaned DataFrame
print(f"Object type: {type(df2)}")
print(f"Shape: {df2.shape}")
print(f"Head:\n{df2.head()}")
```

```python
# Load and display the third dataset
df3 = pd.read_csv(path_3)
print(f"Object type: {type(df3)}")
print(f"Shape: {df3.shape}")
print(f"Head:\n{df3.head()}")
```

```python
pd.read_csv(path_3)[["lat-lon"]].head()
```

```python
# Split the "lat-lon" column on the comma separator and expand into two columns
pd.read_csv(path_3)["lat-lon"].str.split(",", expand=True).head()
```

```python
# Inspect the place_with_parent_names column
(
    pd.read_csv(path_3)
    ["place_with_parent_names"]
    .head()
)
```

```python
# cleaning process using method chaining
df3 = (
    pd.read_csv(path_3)
    # Drop null values from df3
    .dropna()
    # Create "lat" and "lon" columns for df3
    .assign(lat=lambda x: x["lat-lon"]
                .str.split(",", expand=True)[0]           
                .astype(float),
            lon=lambda x: x["lat-lon"]
                .str.split(",", expand=True)[1]           
                .astype(float),
            # Create "state" column for df3
            state=lambda x: x["place_with_parent_names"]
            .str.split("|", expand=True)[2]
           )
    
    # Drop "place_with_parent_names" and "lat-lon" from df3
    .drop(columns=["place_with_parent_names", "lat-lon"])
            
    
)

# Print object type, shape, and head
print("df3 type:", type(df3))
print("df3 shape:", df3.shape)
df3.head()
```

```python
# Check if all column sets are identical 
column_match = set(df1.columns) == set(df2.columns) == set(df3.columns)

if column_match:
    print("All three DataFrames have identical column sets.")
    print(df1.columns.tolist())
    print(df2.columns.tolist())
    print(df3.columns.tolist())
else:
    print("The DataFrames do not have identical column sets.")
    print("df1 vs df2:", set(df1.columns) ^ set(df2.columns))
    print("df2 vs df3:", set(df2.columns) ^ set(df3.columns))
    print("df1 vs df3:", set(df1.columns) ^ set(df3.columns))
```

```python
# Stack the three DataFrames vertically
df = pd.concat([df1, df2, df3], ignore_index=True)

# verify the combined DataFrame
print("Combined shape:", df.shape)
print("Combined type:", type(df))
df.head()
```

```python
# Save the cleaned combined DataFrame to a CSV file for future use
df.to_csv("C:/Dell/Project Github/mexican-real-estate-analysis/data/processed/mexico-real-combined-clean.csv", index=False)
```
