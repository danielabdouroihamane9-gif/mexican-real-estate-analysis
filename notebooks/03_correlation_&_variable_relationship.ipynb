# Notebook 03 - Correlation and Variable Relationships

```python
from pathlib import Path
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv(PROJECT_ROOT / "data/processed/mexico-real-combined-clean.csv")
data["price_per_m2"] = data["price_usd"] / data["area_m2"]
```

```python
pearson = data["area_m2"].corr(data["price_usd"], method="pearson")
spearman = data["area_m2"].corr(data["price_usd"], method="spearman")
print(f"Overall Pearson area-price: {pearson:.4f}")
print(f"Overall Spearman area-price: {spearman:.4f}")
```

```python
grouped = data.groupby("state").agg(area_price_spearman=("price_usd", lambda s: pd.Series(s).corr(data.loc[s.index, "area_m2"], method="spearman")), sample_size=("price_usd", "size")).sort_values("area_price_spearman", ascending=False)
print(grouped.round(4).to_string())
print("\nStates meeting minimum 10 observations:", int((grouped["sample_size"] >= 10).sum()))
```

Grouped correlations include sample counts. States with fewer observations are not interpreted as reliable comparisons. Price-per-square-meter against area is mathematically coupled because area appears in the denominator; a negative result is not independent evidence that larger homes are cheaper.

```python
corr_matrix = data.select_dtypes("number").corr(method="spearman")
fig, ax = plt.subplots(figsize=(9, 6))
sns.heatmap(corr_matrix, annot=True, fmt=".3f", cmap="coolwarm", vmin=-1, vmax=1, ax=ax)
ax.set_title("Spearman Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots()
ax.scatter(data["area_m2"], data["price_per_m2"], alpha=0.25, s=12)
ax.set(title="Area vs. Price per Square Meter", xlabel="Area [sq meters]", ylabel="Price per m2 [USD/m2]")
plt.tight_layout()
plt.show()
```
